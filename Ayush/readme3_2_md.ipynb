# SMS Spam Classification Workflow (Task 2)

I built a Multinomial Naive Bayes classifier to identify SMS spam messages using the SMS Spam Collection dataset.

---

## 1. Exploratory Data Analysis (EDA)

The raw dataset contains 5,574 rows and 2 columns (`label` and `message`).

* **Duplicate Removal:** Identified and removed 414 duplicate entries, resulting in a effective dataset of 5,160 unique messages.
* **Class Distribution:** The cleaned dataset shows a strong class imbalance with 4,518 legitimate messages (`ham`, 87.56%) and 642 spam messages (`spam`, 12.44%).
* **Message Length:** Spam messages are consistently larger on average (137 characters) compared to ham messages (71 characters).

---

## 2. Preprocessing & Feature Extraction

We performed the following text preprocessing steps on every message:
* Converted text to lowercase.
* Stripped out punctuation and special characters.
* Tokenized text into individual words.
* Removed common English stopwords.

Mapped the labels to numerical values (`ham: 0`, `spam: 1`) and split the data into an 80% training set (4,128 samples) and a 20% test set (1,032 samples) using stratified sampling.

We converted the clean text into feature matrices using four distinct extraction strategies:
* **CountVectorizer (Unigrams):** Word counts using single tokens.
* **CountVectorizer (Bigrams):** Word counts using single tokens and adjacent word pairs.
* **TF-IDF Vectorizer (Unigrams):** Term frequencies weighted by inverse document frequency for single tokens.
* **TF-IDF Vectorizer (Bigrams):** Weighted frequencies for single tokens and word pairs.

---

## 3. Model Training & Parameter Experiments

We trained a Multinomial Naive Bayes classifier across all four feature extraction setups. We evaluated five smoothing parameters ($\alpha \in \{0.1, 0.5, 1.0, 2.0, 5.0\}$) for each pipeline:

| Feature Extractor | N-gram Range | Smoothing ($\alpha$) | Accuracy | Precision | Recall | F1-Score | ROC-AUC |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- | :--- |

---

## 4. Key Comparisons

* **CountVectorizer vs. TF-IDF:** CountVectorizer delivers higher recall when using default smoothing ($\alpha=1.0$). However, TF-IDF combined with lower smoothing ($\alpha=0.1$) achieves higher overall precision and top-tier F1-scores because it downweights common non-informative words.
* **Unigrams vs. Bigrams:** Adding bigrams captures phrase context (such as "claim now" or "free entry"), which helped to improve overall precision and reduced false positives.
* **Effect of Alpha ($\alpha$):** Higher $\alpha$ values drastically decreased recall for TF-IDF vectors, whereas a smaller smoothing factor ($\alpha=0.1$) prevents rare spam keywords from getting over-smoothed.

---

## 5. Best-Performing Pipeline & Conclusion

**Optimal Configuration:**
* **Preprocessing:** Standard cleaning + TF-IDF Vectorizer with Unigrams + Bigrams
* **Classifier:** Multinomial Naive Bayes with $\alpha = 0.1$

**Top Performance Metrics:**
* **Accuracy:** 98.26%
* **Precision:** 98.25%
* **Recall:** 87.50%
* **F1-Score:** 0.9256
* **ROC-AUC:** 0.9853

**Why it works:**  
TF-IDF feature extraction downweights common words that appear across all messages while emphasizing distinct spam terms. Bigrams captured the strong context combinations, and setting $\alpha=0.1$ allowed the model to stay sensitive to strong spam signals without over-smoothing probabilities.
