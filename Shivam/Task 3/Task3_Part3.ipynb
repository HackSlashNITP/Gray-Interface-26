```python
import kagglehub
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

path = kagglehub.dataset_download("muratkokludataset/dry-bean-dataset")
print(f"Dataset downloaded to: {path}")

# Constructing the file path directly based on observed structure
# The dataset is typically found in a subdirectory named 'Dry_Bean_Dataset'
# and the file is 'Dry_Bean_Dataset.xlsx'
dataset_subdir = os.path.join(path, 'Dry_Bean_Dataset')
excel_file_name = 'Dry_Bean_Dataset.xlsx'
file_path = os.path.join(dataset_subdir, excel_file_name)

# Verify if the file exists before attempting to read
if os.path.exists(file_path):
    print(f"Found Excel file at: {file_path}")
    df = pd.read_excel(file_path) # Changed to read_excel
    print("\nDataset loaded successfully. Here are the first 5 rows:")
    display(df.head())
else:
    # Provide detailed information if file is still not found
    raise FileNotFoundError(f"Expected Excel file '{excel_file_name}' not found at '{file_path}'.\n" \
                            f"Contents of '{path}': {os.listdir(path)}\n" \
                            f"Contents of '{dataset_subdir}': {os.listdir(dataset_subdir) if os.path.exists(dataset_subdir) else 'Directory not found'}")
```

### Data Preprocessing: Feature-Target Split, Train-Test Split, and Feature Scaling

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import time
import numpy as np

# Separate features (X) and target (y)
X = df.drop('Class', axis=1)
y = df['Class']

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print("\nData split into training and testing sets:")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

# Apply feature scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\nFeatures scaled successfully.")
print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
```

### SVM Model Training and Evaluation

```python
results = []

kernels = ['linear', 'poly', 'rbf', 'sigmoid']

# Hyperparameter ranges to experiment with
C_values = [0.1, 1, 10]
gamma_values = ['scale', 'auto', 0.1, 1]
degree_values = [2, 3, 4] # For Polynomial kernel

for kernel in kernels:
    print(f"\n--- Training SVM with {kernel} kernel ---")
    if kernel == 'poly':
        for C in C_values:
            for gamma in gamma_values:
                for degree in degree_values:
                    print(f"  Kernel: {kernel}, C: {C}, Gamma: {gamma}, Degree: {degree}")
                    start_time = time.time()
                    svm = SVC(kernel=kernel, C=C, gamma=gamma, degree=degree, random_state=42)
                    svm.fit(X_train_scaled, y_train)
                    training_time = time.time() - start_time

                    y_pred = svm.predict(X_test_scaled)

                    accuracy = accuracy_score(y_test, y_pred)
                    precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
                    recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
                    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
                    cm = confusion_matrix(y_test, y_pred)

                    results.append({
                        'kernel': kernel,
                        'C': C,
                        'gamma': gamma,
                        'degree': degree,
                        'training_time': training_time,
                        'accuracy': accuracy,
                        'precision': precision,
                        'recall': recall,
                        'f1_score': f1,
                        'confusion_matrix': cm.tolist() # Convert numpy array to list for easier storage
                    })

    else: # Linear, RBF, Sigmoid
        for C in C_values:
            for gamma in gamma_values:
                print(f"  Kernel: {kernel}, C: {C}, Gamma: {gamma}")
                start_time = time.time()
                svm = SVC(kernel=kernel, C=C, gamma=gamma, random_state=42)
                svm.fit(X_train_scaled, y_train)
                training_time = time.time() - start_time

                y_pred = svm.predict(X_test_scaled)

                accuracy = accuracy_score(y_test, y_pred)
                precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
                recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
                f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
                cm = confusion_matrix(y_test, y_pred)

                results.append({
                    'kernel': kernel,
                    'C': C,
                    'gamma': gamma,
                    'degree': None, # Not applicable for these kernels
                    'training_time': training_time,
                    'accuracy': accuracy,
                    'precision': precision,
                    'recall': recall,
                    'f1_score': f1,
                    'confusion_matrix': cm.tolist()
                })

# Convert results to a DataFrame for easier analysis
results_df = pd.DataFrame(results)

print("\n--- All SVM models trained and evaluated. ---")
display(results_df.sort_values(by='accuracy', ascending=False))

print("\n--- Best performing model based on Accuracy --- ")
best_model = results_df.loc[results_df['accuracy'].idxmax()]
display(best_model)

# Plotting Confusion Matrix for the best model
plt.figure(figsize=(12, 10))
sns.heatmap(best_model['confusion_matrix'], annot=True, fmt='d', cmap='Blues',
            xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
plt.title(f'Confusion Matrix for Best Model (Kernel: {best_model["kernel"]}, C: {best_model["C"]}, Gamma: {best_model["gamma"]}, Degree: {best_model["degree"]})')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```
