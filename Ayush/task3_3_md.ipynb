# Dry Bean Multi-Class Classification Workflow

I Built a model to classify dry bean grains into seven distinct species based on geometric and morphological features using Support Vector Machines (SVM) on the Dry Bean Dataset.

---

## 1. Exploratory Data Analysis (EDA) & Preprocessing

The raw dataset consists of 13,611 rows and 17 attributes (16 numeric feature columns and 1 target class column).

* **Data Cleaning:** There were No missing values across any feature columns. A total of 68 duplicate rows were identified and then removed, leaving 13,543 unique samples.
* **Target Distribution:** The target feature (`Class`) contains 7 dry bean varieties:
  * `DERMASON`: 3,546
  * `SIRA`: 2,636
  * `SEKER`: 2,027
  * `HOROZ`: 1,860
  * `CALI`: 1,630
  * `BARBUNYA`: 1,322
  * `BOMBAY`: 522
* **Splitting & Scaling:** Divided the dataset into an 80% training set (10,834 samples) and a 20% test set (2,709 samples) using stratified sampling. Standard scaling (`StandardScaler`) was applied to scale all features before training.

---

## 4. Confusion Matrix Findings

* **Linear & RBF Kernels:** High precision across distinct categories like `BOMBAY` (103/104 correct predictions).
* **Misclassifications:** There were misclassifications between classes with highly overlapping physical structures, primarily `SIRA` misclassified as `DERMASON` (51 instances) and `BARBUNYA` vs. `CALI`.
* **Sigmoid Kernel:** Showed severe degradation across all target classes, yielding poor overall metrics.

---

## 5. Optimal Model & Conclusion

**Best Performing Configuration:**
* **Kernel:** Radial Basis Function (RBF)
* **Hyperparameters:** $C = 10.0$ or $\gamma = 0.100$
* **Top Performance Metrics:**
  * **Accuracy:** 92.36%
  * **Precision:** 92.40%
  * **Recall:** 92.36%
  * **F1-Score:** 0.9237

**Conclusion:**  
The dry bean dataset consisted of non-linear geometric features (e.g., Aspect Ratio, Roundness, Shape Factors). The RBF kernel maps existing features into a high-dimensional space where non-linear decision boundaries can be separated efficiently. Tuning $\gamma=0.1$ and $C=10$ provides the best trade-off between margin optimization and model flexibility.
