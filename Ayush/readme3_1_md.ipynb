# Wine Quality Classification Workflow (Task 1)

I built a K-Nearest Neighbors (KNN) model to distinguish between red and white wines based on their physical and chemical properties, I used the Wine Quality Dataset.
---

## 1. Exploratory Data Analysis (EDA)

The dataset consist of 6,497 rows and 13 columns (11 numerical features, 1 target quality score, and 1 categorical wine type) .

* **Class Imbalance:** White wines account for ~75% (4,898 samples) of the dataset, while red wines were just ~25% (1,599 samples) .
* **Quality Ratings:** Most wines had a quality rating between 5 and 6 .
* **Feature Distributions:** Chemical features like `residual sugar` and `chlorides` display strong right-skewness with significant outliers .
* **Distinguishing Characteristics:** Scatter plots show clear boundaries between wine types; white wines consistently contains higher residual sugar levels and higher acidity/lower pH compared to red wines .
* **Correlations:** High positive correlation exists between `free sulfur dioxide` and `total sulfur dioxide` (0.72), while strong negative correlation between `density` and `alcohol` (-0.69) .

---

## 2. Preprocessing Steps

* **Missing Values:** Checked missing values across all features(`0 null` count) .
* **Encoding Target:** I mapped the target variable `type` to binary numerical values (`red: 0`, `white: 1`) .
* **Data Splitting:** Applied an 80/20 train-test split stratified by the `type` column to ensure balanced class distributions across splits .
  * **Train set:** 5,197 samples
  * **Test set:** 1,300 samples

---

## 3. Comparative Evaluation & Model Implementation

### Feature Scaling Comparison
As KNN is a distance-based algorithms, it is very sensitive to feature magnitude, three scaling approaches were evaluated using default KNN settings ($K=5$, Euclidean distance, Uniform weights) :

| Scaling Method | Accuracy | F1-Score |
| :--- | :--- | :--- |
| **No Scaling** | 0.950000 | 0.967022 |
| **StandardScaler** | **0.993846** | **0.995918** |
| **MinMaxScaler** | 0.992308 | 0.994893 |

*Observation:* Scaling significantly improved accuracy . **StandardScaler** achieved the highest accuracy and F1-score and was selected for all subsequent steps .

---

### Hyperparameter Tuning ($K$ Values)
Evaluated performance across various values of $K \in [3, 5, 7, 9, 11, 15, 20]$ using Euclidean distance and Uniform weights :

| K Value | Accuracy | F1-Score |
| :--- | :--- | :--- |
| 3 | 0.993846 | 0.995918 |
| 5 | 0.993846 | 0.995918 |
| **7** | **0.995385** | **0.996939** |
| **9** | **0.995385** | **0.996939** |
| 11 | 0.994615 | 0.996430 |
| 15 | 0.992308 | 0.994898 |
| 20 | 0.992308 | 0.994893 |

*Observation:* Performance peaks at $K=7$ and $K=9$ . Increasing $K$ beyond 11 causes performance to degrade due to over-smoothing . $K=7$ was chosen as optimal .

---

### Distance Metrics Comparison ($K=7$, Uniform Weights)

| Distance Metric | Accuracy | F1-Score |
| :--- | :--- | :--- |
| **Manhattan** | **0.995385** | **0.996942** |
| **Euclidean** | 0.995385 | 0.996939 |
| **Minkowski ($p=3$)** | 0.994615 | 0.996427 |

*Observation:* **Manhattan distance** gives a slightly superior F1-Score compared to Euclidean distance .

---

### Weighting Methods Comparison ($K=7$, Manhattan Distance)

| Weighting Method | Accuracy | F1-Score |
| :--- | :--- | :--- |
| **Uniform** | **0.995385** | **0.996942** |
| **Distance** | 0.994615 | 0.996427 |

*Observation:* **Uniform weighting** slightly outperforms distance-based weighting .

---

## 4. Key Observations & Final Model Summary

* **Optimal Configuration:**
  * **Scaling:** `StandardScaler`
  * **$K$ Neighbors:** `7`
  * **Distance Metric:** `Manhattan`
  * **Weighting Scheme:** `Uniform`

* **Final Model Performance:**
  * **Accuracy:** `99.54%` (0.995385)
  * **F1-Score:** `99.69%` (0.996942)

**Conclusion:**  
Feature scaling is critical for KNN models when working with continuous data with differing scales . `StandardScaler` standardized feature ranges effectively . Using **Manhattan distance** provided optimal separation in high-dimensional space, and $K=7$ struck the best balance between variance and bias .
